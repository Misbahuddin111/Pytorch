```python
import torch
x = torch.tensor(3.0,requires_grad=True)
print(x)
```

```python
# simple forward computation
import torch
x = torch.tensor(3.0 ,requires_grad=True)
y = x**2
print(y)
# powerBackward0 means pytorch remembers y = x**2 
```

```python
# backward propagation
x = torch.tensor(3.0,requires_grad=True)
y = x**2
y.backward()
print(x.grad)
```

```python
# viewing gradinent 
import torch
x = torch.tensor(4.0,requires_grad=True)
y= x**2
y.backward()
print(x.grad)
```

```python
# clearning gradinet 
import torch
x = torch.tensor(4.0,requires_grad=True)
y= x**2
y.backward()
print(x.grad)
x.grad.zero_()
```

```python
# multiple variabales
# z = x**2 + y**2
x = torch.tensor(2.0,requires_grad=True)
y = torch.tensor(3.0,requires_grad=True)
z = x**2 + y**2
z.backward()
print(x.grad)
print(y.grad)
```

```python
# vectors gradients
x = torch.tensor([1.,2.,3.,],requires_grad=True)
y = x * 2 
y.sum().backward()
print(x.grad)
```

```python
# torch.no_grad()
import torch
x = torch.tensor(2.0,requires_grad=True)
with torch.no_grad():
    y = x * 3
    print(y)
```

```python
# detech()
x = torch.tensor(3.0, requires_grad=True)
y = x*2
z = y.detach()
print(z.requires_grad)
```

```python
# requires_grad
x = torch.tensor(5.0)
x.requires_grad_(True)
print(x.requires_grad)
```

```python
# checking requires_grad
x = torch.tensor(10.0,requires_grad=True)
print(x.requires_grad)
```

```python
# neural network example
w = torch.tensor(2.0,requires_grad=True)
x = torch.tensor(3.0)
y_true = torch.tensor(10.0)
y_pred = w*x
loss = (y_pred - y_true)**2 
loss.backward()
print(w.grad)
```

```python
# optimizer used autogrd
w = torch.tensor(1.0,requires_grad=True)
optimizer = torch.optim.SGD([w],lr=0.01)
for epoch in range(100):
    y_pred = w*2
    loss = (y_pred - 8) ** 2
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
print(w)
```
