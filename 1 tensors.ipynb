```python
import torch
```

```python
# 0 D tensor
x = torch.tensor(5)
print(x.ndim)

# matrix 
m = torch.tensor([[1,2,3],[4,5,6]],dtype=torch.float32)
print("matrix tensor Dimension:",m.ndim)
print("shape of tensor:",m.shape)
print(m)
print(m.dtype)
```

###  Special Tensor Creation

```python
ones = torch.ones(2,3,dtype=torch.int64)
print("ones:\n",ones)
# zeros 
zeros = torch.zeros(2,2)
print("zeros:\n",zeros)
# random numbers
randd = torch.rand(3,3)
print("randoms:\n",randd)
# range tensor
y = torch.arange(1,10)
print("range:",y)
# reshape
print("reshape:\n",y.reshape(3,3))
# indexing
z = torch.tensor([
    [10, 20, 30],
    [40, 50, 60]
])
print(z[0])
print(z[1][1])
# slicing
x = torch.tensor([1, 2, 3, 4, 5])

print(x[1:4])
print(x[0:3])
```

```python
# Arthimatic operations
x = torch.tensor([1, 2, 3])
y = torch.tensor([4, 5, 6])
print("addition:\n",x+y)
print("subtraction:\n",x-y)
print("multiplication:\n",x*y)
print("division:\n",x*y)
print("power:\n",x**2)

meann = torch.tensor([1, 2, 3, 4], dtype=torch.float32)
print("mean:\n",torch.mean(meann))
s = torch.tensor([1, 2, 3, 4])

print("sum:\n",torch.sum(s))
# maximum value
maax = torch.tensor([5, 2, 9, 1])

print("max:\n",torch.max(maax))
minn = torch.tensor([5, 2, 9, 1])

print("min:\n",torch.min(minn))
# armax
x = torch.tensor([5, 2, 9, 1])

print("argmax:\n",torch.argmax(x))
x = torch.tensor([5, 2, 9, 1])

print("argmin:\n",torch.argmin(x))
```
